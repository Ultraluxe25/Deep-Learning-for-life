CNN позволяют обнаруживать объект вне зависимости положения на картинке.

Padding удерживает текущее разрешение изображения, при ядре свертки 3 на 3.



Каналов на следующем слои может быть столько, сколько фильтров (ядер свертки) мы применяем к текущему слою.

Пулинг слой применяется отдельно к каждому каналу.

### Пулинг понижает размерность весов модели, но при этом сохраняет важную информацию на слое.
