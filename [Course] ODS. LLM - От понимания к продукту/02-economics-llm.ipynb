**Инференс (inference)** в контексте больших языковых моделей (LLM) — это процесс, когда уже обученная модель используется для генерации ответов, предсказаний или текста на основе нового запроса пользователя.

Деньги уходят не только на входящие токены, но еще и на риазонинг, который может занимать в 100 раз больше токенов чем вход.
